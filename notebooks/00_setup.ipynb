```python
import os
from dotenv import load_dotenv
load_dotenv("../.env")

from langchain_groq import ChatGroq

llm = ChatGroq(model=os.getenv("GROQ_MODEL"), temperature=0)
resp = llm.invoke("What is the diference between an armlock and a triangle?")
print(resp.content)
```

```python
import os
import requests

url = "https://www.googleapis.com/youtube/v3/search"
params = {
    "part": "snippet",
    "q": "fuga armlock guarda fechada",
    "type": "video",
    "maxResults": 1,
    "key": os.getenv("YOUTUBE_API_KEY"),
}
r = requests.get(url, params=params, timeout=10)
r.raise_for_status()

item = r.json()["items"][0]
titulo = item["snippet"]["title"]
video_id = item["id"]["videoId"]
print(titulo)
print(f"https://www.youtube.com/watch?v={video_id}")
```

```python
url = "https://en.wikipedia.org/api/rest_v1/page/summary/Armbar"
headers = {"User-Agent": "bjj-agent/0.1 (estudo)"}

r = requests.get(url, headers=headers, timeout=10)
r.raise_for_status()

dados = r.json()
print(dados["title"])
print(dados["extract"])
```

```python
import sqlite3
import pandas as pd

df = pd.read_csv("../data/raw/adcc_matches.csv")
print(df.shape)

conn = sqlite3.connect("../data/db/bjj.sqlite")
df.to_sql("lutas", conn, if_exists="replace", index=False)

total = pd.read_sql("SELECT COUNT(*) AS total_lutas FROM lutas", conn)
print(total)
conn.close()
```
